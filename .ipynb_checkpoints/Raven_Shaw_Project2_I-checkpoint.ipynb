#### **Section 2: Data Cleaning and Indexing**

```python
df_new.duplicated().sum() #checking for duplicates
```

```python
df_drop_dup=df_new.copy()
df_drop_dup.drop_duplicates(inplace=True) #now only have 252 rows demonstrating that the duplicated rows were dropped
```

```python
df_drop_dup.fillna({'Mid-Career 10th Percentile Salary': df_new['Mid-Career 10th Percentile Salary'].median(),
               'Mid-Career 90th Percentile Salary': df_new['Mid-Career 90th Percentile Salary'].median()}, inplace=True) 
#filling in missing values in both 10th and 90th percentile columns with median; using inplace=True to ensure changes stick
df_drop_dup #displaying edited data 
```

```python
df_region_type = df_drop_dup.copy() #equating new copy to "df_region_type"
type_mode = df_region_type['School Type'].mode()[0] #selecting most frequent category
df_region_type['School Type']=df_region_type['School Type'].fillna(type_mode)  #using mode imputation for "School Type" values 
df_region_type #printing changes
```

```python
df_region_type['Region']=df_region_type['Region'].fillna('Unknown') #replacing missing values in region with "Unknown"
df_region_type #displaying the data
```

```python
df_region_type.isnull().sum() #rechecking the null values
```

***Section 2 Explanation***: For this section, I focused on replacing missing values. I started by removing the 5 duplicates. As aforementioned in section 1, I replaced the missing values in the "region," "school type," and "10th and 90th percentile columns." Nevertheless, I used different methods for each. For the region I replaced using the word "Unknown" because region is systematic data whereas for school type I could rely on mode. For the numerical columns, I focused on replacing missing values using median. I then rechecked the null values to ensure they are all fill in. I then saved the cleaned dataset (see below).

*Interesting Takeaway*: There is a disproportionately large difference between the Mid-Career 90th Percentile group mean and Mid-Career 10th Percentile group mean which is something I want to explore further using descriptive analysis.

```python
salary_by_type1 = df_cleaned.groupby('School Type')['Starting Median Salary'].median() 
# calculating the median starting salary for each school type in the cleaned dataset
salary_by_type1
```

```python
salary_by_type2 = df_cleaned.groupby('School Type')['Mid-Career Median Salary'].median()
# calculating the median mid career salary for each school type in the cleaned dataset
salary_by_type2
```

```python
salary_by_region1 = df_cleaned.groupby('Region')['Mid-Career 10th Percentile Salary'].mean()
#calculating the mean salary for each region in the 10th percentile
salary_by_region1
```

```python
salary_by_region2 = df_cleaned.groupby('Region')['Mid-Career 90th Percentile Salary'].mean()
#calculating the mean salary for each region in the 90th percentile
salary_by_region2
```

```python
salary_inequality = salary_by_region2 - salary_by_region1 #subtracting the difference between mean values of the 90th and 10th percentiles
salary_inequality
```

```python
def classify_status(salary): #feature engineering beginning with creating a user defined function
    if pd.isnull(salary):
        return 'Unknown'
    elif salary < 30000:
        return 'Lower Class'
    elif salary < 94000:
        return 'Middle Class'
    elif salary < 153000:
        return 'Upper Middle Class'
    else:
        return 'Upper CLass'
df_cleaned=df_cleaned.copy()
df_cleaned['Median Socioeconomic Class']=df_cleaned['Mid-Career Median Salary'].apply(classify_status)
#creating a new column "Median Socioeconomic Class" based on defined function above
df_cleaned
```

```python
df_cleaned.to_csv('Raven_Shaw_project2_part1_data_cleaned.csv') #saving the cleaned CSV
```

***Section 3 Explanation***: This section was based on descriptive statistics. After revieiwng the general statistics (based on the function "describe"), I first looked at the value counts for the categorical columns ("region" and "school type"). I then began grouping different columns (using the "groupby" function) to gain a deeper insight into the data provided. I calculated the median starting and midcareer salary for each school type in the cleaned dataset. I also calculated the mean salary for each region in the 10th and 90th percentile groups. To understand more indepthly the potential for income inequality, I substracted these values from each other. Lastly, using feature engineering I used a user defined function to calculate the economic classes based on the salaries (information I got from the Census' Bureau). From there I created a new column based on this information. Finally, I saved the final, cleaned and refined version of the dataset.
