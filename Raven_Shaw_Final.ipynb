```python
#using dictionary function to assign countries to their alpha-3 code 
country_code = {
    'USA' : 'USA',
    'Canada' : 'CAN',
    'Switzerland' : 'CHE',
    'Sweden' : 'SWE',
    'Netherlands': 'NLD',
    'Russia': 'RUS',
    'South Africa': 'ZAF',
    'Mexico' : 'MEX',
    'South Korea' : 'PRK',
    'Brazil' : 'BRA',
    'Japan' : 'JPN',
    'China' : 'CHN',
    'India' : 'IND',
    'Australia' : 'AUS',
    'Spain' : 'ESP',
    'Italy' : 'ITA',
    'France' : 'FRA',
    'Germany' : 'DEU',
    'UK' : 'GBR',
    'UAE' : 'ARE'
}
df_cleaned['Code']=df_cleaned['Country'].map(country_code) #creating new "code" column
```

```python
#using dictionary function to create regions based on countries
region_creation = {
    'USA' : 'North America',
    'Canada' : 'North America',
    'Switzerland' : 'Europe',
    'Sweden' : 'Europe',
    'Netherlands': 'Europe',
    'Russia': 'Europe',
    'South Africa': 'Africa',
    'Mexico' : 'North America',
    'South Korea' : 'Asia',
    'Brazil' : 'South America',
    'Japan' : 'Asia',
    'China' : 'Asia',
    'India' : 'Asia',
    'Australia' : 'Oceania',
    'Spain' : 'Europe',
    'Italy' : 'Europe',
    'France' : 'Europe',
    'Germany' : 'Europe',
    'UK' : 'Europe',
    'UAE' : 'Asia'
}
df_cleaned['Region'] = df_cleaned['Country'].map(region_creation) #creating new "Region" column
df_cleaned #displaying revised dataset
```

#### **Section 4: Data Visualizations**

```python
df_2020 = df_cleaned[df_cleaned["Year_cat"] == "2020"] #isolating dataset to solely the years 2020 and 2023
df_2023 = df_cleaned[df_cleaned["Year_cat"] == "2023"]
```

```python
numeric_df = df_cleaned.select_dtypes(include='number') # Only selecting numeric columns

plt.figure(figsize=(10, 6)) #setting size of graph
sns.heatmap(numeric_df.corr(), annot=True, cmap='autumn_r') #creating heatmap
plt.title("Correlation Matrix (Seaborn)") #label for title
plt.show() #display graph
```

*Analysis*: Evidently, the heat map shows that there aren't any significantly strong correlations across the different factors. However, rent index and house price index, inflation rate and GDP growth, and population and rent index have relatively stronger correlatons (albeit differ in whether those correlations are positive and negative). Nevertheless, this heatmap provides crucial guidance in how I continue my dataset.

```python
plt.bar(house_indict2.index, house_indict2.values, color='gold') #constructing plot bar
plt.title("Average House Price Index by Country (Matplotlib - plt.bar)") #label title
plt.xlabel("Country") #label x-axis
plt.ylabel("Average House Price Index") #label y-axis
plt.xticks(rotation=45) #setting ticks of x-axis at an angle for easier readability
plt.tight_layout() #function that adjusts spacing
plt.show() #display graph
```

*Analysis*: Using the groupby function of "house_indict2" I was able to utilize the mean values to implement within this barplot to demonstrate that the top three countries with highest average House Price Index are Sweden, Netherlands and Brazil. Although Brazil is the outlier in terms of region, this trend overall implies a potential connection with government policy and oversight (or lackthereof) within their housing markets.

```python
plt.hist(df_cleaned['Mortgage Rate (%)'], bins=40, edgecolor='black', color='orange') #creating histogram
plt.xlabel("Mortgage Rate") #label for x-axis
plt.ylabel("Frequency") #label for y-axis
plt.title("Mortgage Rate Distribution") #titling of graph
#plt.savefig('Raven_Shaw_Project2_Histogram.jpg') #saving graph as a jpg on laptop
plt.show() #display graph
```

*Analysis*: This histogram displays the distribution fo mortgage rate (%), and although there is no particular shape of the graph, except perhaps "normal," there are still outliers with heavy influence on the overall data. There are also gaps in the data which is an important note. Overall the highest peak at above 10% suggests high interests rates in a particularly concentrated area. 

```python
plt.scatter(df_cleaned['House Price Index'], df_cleaned['Rent Index'], color='gold')
# Add labels and title
plt.title("House Price Index vs Rent Index") #label title
plt.xlabel("House Price Index")
plt.ylabel("Rent Index")

plt.grid(True)
plt.show()
```

*Analysis:* This scatter plot demonstrates the relationship between house price index and rent index, however it is evident that due to the lack of pattern between the data points, there is no relationship between these two columns/factors.

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_cleaned, x='Year', y='Affordability Ratio', hue='Region', marker='o', ci=None)

plt.title("Affordability Ratio Over Time by Region")
plt.xlabel("Year")
plt.ylabel("Affordability Ratio")
plt.legend(title='Region')
plt.grid(True)
plt.tight_layout()
plt.show()
```

*Analysis:* This lineplot demonstrates the change in affordability ratio over time by region. Evidently, the plot fluctuates across time regardless of regions. Africa has the largest fluctuations and Asia changes in the smallest increments.

```python
fig = px.choropleth(df_cleaned,
                     locations="Code",
                     color="House Price Index",
                     hover_name="Country",
                     animation_frame="Year",
                     projection="natural earth", #many options such as "orthographic", "equirectangular", "natural earth", "mercator". "miller". "robinson "
                     title="Average House Price Index by Country",
                     color_continuous_scale="thermal", #many options such as viridis, plasma, hot, ice, blues, reds, etc
                     range_color=(0,40),
                     height=600)
fig.update_layout(geo=dict(showframe=False, showcoastlines=True))
fig.show()
```

*Analysis:* This geomap depicts the average house price index across the countries. Evidently, regardless of country they all have a high house price index.
